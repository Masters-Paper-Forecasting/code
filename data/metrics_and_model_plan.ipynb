This notebook explains how we'll judge if our air quality prediction models are doing a good job and our plan for getting them ready.



We're using six different ways to check how well our models predict PM2.5 levels. Each one tells us something unique about the model's accuracy. By looking at all six, we get a full and honest picture of how good our predictions are.

## 1. MAE — Mean Absolute Error

MAE tells us, on average, how much our predicted PM2.5 numbers differ from the actual levels. Think of it as the typical 'miss' size, whether the model predicted too high or too low. We add up all these 'misses' (ignoring if they were positive or negative) and then divide by the total number of predictions.

**Example:** If the actual PM2.5 is 12 and our model guesses 15.5, that's a 3.5 difference. If, on average, all our predictions are about 3.2 units off, that's our MAE. For comparison, the World Health Organization (WHO) suggests a daily limit of 15 units. So, a 3.2 MAE means we're roughly 21% off compared to that limit, which helps us understand how significant the error is.

We use MAE because it's easy to explain to anyone, even those without a technical background, like government officials or health experts.

**Formula:** MAE = (1/number of predictions) × Sum of |actual value - predicted value|

**A smaller number is better.**

## 2. RMSE — Root Mean Squared Error  

RMSE is similar to MAE, but it really highlights big prediction errors. Imagine errors as steps: MAE counts each step the same. RMSE makes bigger steps count much more. This is because it squares each error before averaging them, and then takes the square root at the end.

**Example:** If our model predicts 14 when it should be 45 (a huge mistake, maybe during a dust storm), that single error, when squared, becomes very large (961). A typical small mistake of 3 units only becomes 9 when squared. So, RMSE is extra sensitive to those days with extreme pollution, which is important for air quality, as large errors on bad air days are more serious for public health.

RMSE will always be the same as or bigger than MAE. If they're nearly the same, our model makes pretty consistent errors. If RMSE is much higher than MAE, it means our model made a few very large errors, and we should investigate why.

**Formula:** RMSE = Square Root of [(1/number of predictions) × Sum of (actual value - predicted value)²]

**A smaller number is better.**

## 3. R² — R-Squared  

R² tells us how well our model explains the ups and downs in actual PM2.5 levels. It's usually a number between 0 and 1. If it's 1, our model perfectly explains all the changes. If it's 0, it explains none of them. Sometimes it can even be negative, meaning our model is worse than just guessing the average PM2.5 every time.

**Example:** If PM2.5 levels vary from 4 on a clean day to 45 during heavy pollution, an R² of 0.85 means our model captures 85% of what causes that variation. The other 15% is still a mystery. Most good PM2.5 prediction models (like those using LSTM) usually have R² values between 0.75 and 0.95.

R² is great for quickly comparing different models. If one model has an R² of 0.78 and another has 0.89, it's clear the second one is much better at understanding the underlying patterns.

**Formula:** R² = 1 - [Sum of (actual value - predicted value)² / Sum of (actual value - average actual value)²]

**A higher number is better. The best is 1.0.**

## 4. MAPE — Mean Absolute Percentage Error  
MAPE looks at each error as a percentage of the actual value. For example, if the real PM2.5 is 10 and our model says 13, that's a 30% error for that day. MAPE then averages all these daily percentage errors.

The cool thing about MAPE is that it's useful for comparing predictions even if the actual numbers are very different. An MAPE of 15% means our model is typically 15% off, no matter if the PM2.5 levels are high or low.

One small catch: MAPE can get wacky if the actual PM2.5 value is very close to zero, because dividing by a tiny number makes the percentage huge. But for London Marylebone Road, PM2.5 rarely drops below 2, so this shouldn't be a big problem for us.

**Formula:** MAPE = (100 / number of predictions) × Sum of |(actual value - predicted value) / actual value|

**A smaller number is better.**

## 5. MSE — Mean Squared Error  

MSE is basically RMSE before we take the square root. We report it because it's the number that many models (especially neural networks like LSTMs) try to make as small as possible when they're learning. By showing MSE, we make it clear that how we train the model aligns with how we're evaluating it.

**Example:** If our RMSE is 4.8, then our MSE would be 23.04. The units of MSE are 'squared' (e.g., micrograms squared per cubic meter squared), which makes it a bit harder to understand on its own. But reporting it with RMSE helps keep our evaluation consistent with what's commonly seen in other research papers on PM2.5 forecasting.

**Formula:** MSE = (1/number of predictions) × Sum of (actual value - predicted value)²

**A smaller number is better.**

## 6. MBE — Mean Bias Error  
MBE tells us if our model tends to consistently guess too high or too low. We calculate it by averaging all the prediction errors, but this time we *do* keep track of whether the error was positive (predicted too high) or negative (predicted too low).

**Example:** If our MBE is -2.1, it means, on average, our model is predicting PM2.5 levels 2.1 units lower than they actually are. In terms of public health, this is a dangerous kind of bias. If a model always underpredicts pollution, authorities might not issue health warnings or traffic restrictions when they should.

On the other hand, a positive MBE means the model is usually over-cautious. This is less harmful to health but could make people lose trust if warnings are given too often without real need. MBE helps us see if our model has a systematic 'lean' in one direction, something MAE and RMSE can't tell us on their own.

**Formula:** MBE = (1/number of predictions) × Sum of (predicted value - actual value)

**The closer to zero, the better. A negative value means it tends to guess too low. A positive value means it tends to guess too high.**

## Why These Metrics Matter for Air Quality (PM2.5) Prediction

Understanding air quality is critical for public health. These metrics aren't just abstract numbers; they directly relate to how well we can protect people from harmful pollution.

### MAE (Mean Absolute Error): Essential for General Risk Assessment

For PM2.5, MAE gives us a straightforward understanding of the average daily exposure error. If our MAE is, for example, 3 µg/m³, it means on an average day, our prediction might be off by about 3 micrograms per cubic meter. This directly impacts public health advice. If the actual PM2.5 is 20 and we predict 17, or vice-versa, the health impact changes. Knowing the average deviation helps health authorities understand the general reliability of our forecasts and how much buffer they might need to add to warnings.

### RMSE (Root Mean Squared Error): Crucial for Identifying Extreme Pollution Events

RMSE's sensitivity to large errors is invaluable in air quality. High PM2.5 levels, even for short periods, can trigger asthma attacks, heart problems, and other severe health issues. An accurate prediction of a pollution 'spike' is far more important than accurately predicting a perfectly clean day. If our model makes a large error during a high pollution event (e.g., predicting 10 when it's actually 40), RMSE penalizes this heavily. This encourages our models to be more accurate when it matters most – during dangerous pollution episodes.

### R² (R-Squared): Understanding the Model's Explanatory Power

R² tells us how much of the variation in PM2.5 levels our model can actually account for. In environmental science, PM2.5 concentrations are influenced by a complex interplay of weather, traffic, industrial emissions, and even transboundary pollution. A high R² suggests our model is capturing these complex relationships effectively. If R² is low, it means there are significant factors affecting air quality that our model isn't seeing, which could lead to missed predictions, especially for novel pollution events or under-represented conditions.

### MAPE (Mean Absolute Percentage Error): Useful for Relative Accuracy Comparisons

MAPE is particularly useful for air quality because it provides a relative sense of error. A 10% error when PM2.5 is 5 µg/m³ (an error of 0.5 µg/m³) is very different from a 10% error when PM2.5 is 50 µg/m³ (an error of 5 µg/m³). MAPE treats these proportionally, allowing us to gauge how consistently accurate our model is across different pollution levels, which can be valuable for setting adaptive warning thresholds.

### MSE (Mean Squared Error): Direct Link to Model Optimization

MSE is fundamentally tied to how many machine learning models learn. When we train a model to predict PM2.5, its internal algorithms often try to minimize MSE. By reporting MSE, we ensure transparency and alignment between our model's learning objective and our evaluation. While its units are less intuitive, it provides a direct measure of the model's 'cost function' and allows for easier comparison with academic research that often uses MSE as a primary optimization target.

### MBE (Mean Bias Error): Identifying Systemic Over or Under-Prediction

MBE is perhaps one of the most health-critical metrics for air quality. A consistent negative MBE means our model is routinely *underestimating* PM2.5 levels. This is a severe public health risk: if official warnings are based on such a model, people might be unknowingly exposed to higher-than-predicted pollution, failing to take protective measures. Conversely, a consistent positive MBE might lead to unnecessary warnings, which could cause public distrust. MBE helps us tune our models to be as unbiased as possible, ensuring that the predicted risk aligns closely with the actual risk.

# Part 2 — Our Plan to Train the Models

We have three team members, and each will work on two different prediction models. Here's how we're splitting up the work:

## Team Member 1 — Random Forest and XGBoost

**Random Forest** is like having many independent 'decision-makers' (decision trees) all vote on the prediction, and then we take the average. Each decision-maker learns from a slightly different part of the data, which helps it avoid being overly specific to one situation (overfitting). For predicting PM2.5, it's good at finding complex connections between things like yesterday's pollution, the day of the week, or other air chemicals, even without understanding the flow of time directly. It also tells us which pieces of information are most important for its predictions. Because of this, it's our starting point – the model we'll compare all others against.

**XGBoost** is a different kind of 'team' approach. Instead of learning independently, it builds its decision-makers one by one, with each new one trying to fix the mistakes of the previous ones. This focused learning usually makes XGBoost perform even better than Random Forest. It's also highly adjustable. By comparing XGBoost with Random Forest, we can see if learning in a more structured, corrective way (boosting) works better than just averaging many independent guesses (bagging) for our air quality predictions.

## Team Member 2 — LSTM and GRU

**LSTM** (Long Short-Term Memory) is a special kind of brain-like network designed to understand patterns in sequences, like daily PM2.5 readings. It has clever 'gates' that decide what past information to remember or forget. This allows it to recall important details from days or even weeks ago, which is crucial for understanding how pollution builds up over time. For our project, it would typically look at a recent window of PM2.5 and other related values (e.g., the last 7 to 30 days) to predict future levels.

**GRU** (Gated Recurrent Unit) is a simpler version of LSTM. It has fewer 'gates,' making it a bit faster to train and less likely to get confused by smaller datasets. Often, GRU and LSTM perform similarly. By including both, we can figure out if the extra complexity of LSTM really gives us better predictions for our specific task, or if GRU is good enough.

## Team Member 3 — CNN-LSTM and Transformer

**CNN-LSTM** is a hybrid model that combines two powerful techniques. First, a **CNN** (Convolutional Neural Network) part scans the recent history to spot quick, local patterns – like a sudden spike in PM2.5 over a few days. Then, an **LSTM** part takes these spotted patterns and figures out how they fit into the longer-term trend. This combination is great because it can catch both short-term pollution events and the overall long-term flow of air quality. Recent research shows that this approach often beats standalone LSTMs for PM2.5 forecasting.

**Transformer** models are a newer kind of network that don't process information step-by-step like LSTMs. Instead, they look at all past time points at once and figure out which ones are most important for predicting the future. This makes them particularly good for predicting far into the future, as they can directly connect a future prediction to relevant events from many days ago. New studies demonstrate that Transformer-based models are very effective for PM2.5 forecasting, especially when we need to predict well in advance, which is a key goal for our project.

## Alternatives if Computational Resources are Limited

If the computational demands of models like the Transformer or even complex CNN-LSTMs become a bottleneck, especially given their need for large datasets and longer training times, here are some viable alternatives that still offer strong performance for PM2.5 forecasting:

### 1. ARIMA/SARIMA (AutoRegressive Integrated Moving Average / Seasonal ARIMA)

*   **Description:** These are classic statistical models for time series forecasting. ARIMA handles non-seasonal time series, while SARIMA extends it to account for seasonal patterns, which are very common in air quality data (e.g., daily, weekly, or yearly cycles).
*   **Strengths:**
    *   **Low Computational Cost:** Significantly less demanding to train and deploy compared to deep learning models.
    *   **Interpretability:** Parameters often have a clear statistical meaning, allowing for better understanding of the underlying time series dynamics.
    *   **Good for Shorter Time Horizons:** Can provide reliable forecasts for short to medium-term predictions.
    *   **Requires Less Data:** Can perform well even with smaller historical datasets.
*   **Weaknesses:**
    *   **Limited in Non-Linearity:** Struggles with highly non-linear relationships and complex interactions between variables, which are common in PM2.5.
    *   **Feature Engineering:** Requires careful manual feature engineering (e.g., adding external regressors like temperature or wind speed).
    *   **Doesn't Scale Well with Many Features:** Primarily designed for univariate time series or a few exogenous variables.



### 2. LightGBM / CatBoost (Gradient Boosting Decision Trees)

*   **Description:** These are highly optimized versions of Gradient Boosting Decision Trees, similar to XGBoost but often faster and more memory-efficient. They can be effectively used for time series by engineering lagged features (e.g., PM2.5 from 1 day ago, 7 days ago, etc.) and other time-based features (day of week, month, year, holiday flags).
*   **Strengths:**
    *   **High Performance:** Often achieve state-of-the-art results on tabular data and well-engineered time series tasks.
    *   **Speed & Efficiency:** Designed for fast training and low memory usage, making them good for resource-constrained environments.
    *   **Handles Non-Linearity & Interactions:** Excellent at modeling complex relationships between features.
    *   **Feature Importance:** Can provide insights into which features are most influential for predictions.
*   **Weaknesses:**
    *   **Requires Manual Feature Engineering:** Effective time series forecasting with these models heavily relies on creating appropriate lagged and temporal features.
    *   **Less Suited for Very Long Sequences:** Without deep feature engineering, they don't inherently capture long-term dependencies as well as LSTMs or Transformers.
    *   **Prone to Overfitting:** Requires careful tuning of hyperparameters to avoid overfitting.
